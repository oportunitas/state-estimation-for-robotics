Given two PDFs (over the same random variable), $p_1(\mathbf{x})$ and $p_2(\mathbf{x})$, we may be interested in quantifying how 'far apart' they are. One method is to compute the *Kullback-Leibler (KL) divergence* (Kullback and Leibler, 1951) between the two:

$$\mathrm{KL}(p_2||p_1) = - \int p_2(\mathbf{x}) \ln \left( \frac{p_1(\mathbf{x})}{p_2(\mathbf{x})} \right) d\mathbf{x} \geq 0. \tag{2.25}$$

```python
def _kl_divergence(p2, p1):
  bounds = list(zip(
    [min(b[0] for b in dims) for dims in zip(p1.bounds, p2.bounds)],
    [max(b[1] for b in dims) for dims in zip(p1.bounds, p2.bounds)]
  ))
  foo = _ContinuousFunction(
    f=(lambda x: (
      p2(x) * math.log(r) if (p2(x) > 0 and (r := p1(x) / p2(x)) > 0) else 0.0
    )),
    bounds=bounds
  )

  return foo.integrate() * -1
```

```python
if __name__ == '__main__' and '__file__' not in globals():
  p1 = _ContinuousFunction(f=(_gaussian(0.0, 1.0)), bounds=[(-128.0, 128.0)])
  p2 = _ContinuousFunction(f=(_gaussian(2.0, 5.0)), bounds=[(-128.0, 128.0)])

  print(_kl_divergence(p2, p1))
```
